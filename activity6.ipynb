# National Analysis

Visualizing where perpetrators of a specific nationality carried out attacks.

```python
filter1 = df[df['natlty1_txt'] == 'Iraqi']
filter2 = filter1[filter1['iyear'] > 2000]
filter3 = filter2[filter2['attacktype1_txt'] == 'Bombing/Explosion']  # You can change this to include all if needed

final_filter = filter3.groupby(['latitude','longitude']).size().reset_index(name='incident_count')

heatmap_data = final_filter[['latitude','longitude','incident_count']].values.tolist()

nationality_map = folium.Map(location=[30, 45], zoom_start=4)
HeatMap(heatmap_data, radius=8).add_to(nationality_map)

nationality_map.save('heatmap_nationality_iraqi.html')
nationality_map
```

# INSIGTH 3
This heatmap shows the geographical distribution of terrorist attacks carried out by Iraqi perpetrators after 2000. It highlights regions that have seen significant terrorist activity attributed to individuals of Iraqi nationality, offering insights into the international scale and impact of these attacks.

# Regional Analysis
Where are attacks happening most frequently by region?

```python
filter1 = df[df['region_txt'] == 'Middle East & North Africa']
filter2 = filter1[filter1['iyear'] > 2000]
filter3 = filter2[filter2['attacktype1_txt'] == 'Bombing/Explosion']

final_filter = filter3.groupby(['latitude','longitude']).size().reset_index(name='incident_count')

# Heatmap
from folium.plugins import HeatMap
import folium

heatmap_data = final_filter[['latitude','longitude','incident_count']].values.tolist()

region_map = folium.Map(location=[25, 45], zoom_start=4)
HeatMap(heatmap_data, radius=8, max_zoom=13).add_to(region_map)

region_map.save('heatmap_region_mena_bombings.html')
region_map
```

# INSIGTH 4
This heatmap visualizes the concentration of bombing and explosion attacks in the Middle East and North Africa since the year 2000. The map clearly highlights the intensity of such incidents in the region, emphasizing the ongoing challenges faced by countries in this part of the world due to terrorism.

# Target Analysis
Where certain types of targets are attacked (e.g., civilians, government, police)

```python
filter1 = df[df['targtype1_txt'] == 'Private Citizens & Property']
filter2 = filter1[filter1['iyear'] > 2000]
filter3 = filter2[filter2['attacktype1_txt'] == 'Bombing/Explosion']  # You can remove this line to include all attack types

final_filter = filter3.groupby(['latitude','longitude']).size().reset_index(name='incident_count')

heatmap_data = final_filter[['latitude','longitude','incident_count']].values.tolist()

target_map = folium.Map(location=[20, 0], zoom_start=2)
HeatMap(heatmap_data, radius=8).add_to(target_map)

target_map.save('heatmap_target_civilians.html')
target_map
```

# INSIGTH 5 
This map demonstrates the geographical areas where civilians (private citizens & property) have been the primary targets of terrorist attacks since 2000. The heatmap underscores the vulnerability of civilian populations around the world, particularly in regions affected by terrorism and violent conflicts.

# Casualty Analysis
 Fatal Attacks (nkill > 0) After 2000

```python
filter1 = df[df['nkill'] > 0]
filter2 = filter1[filter1['iyear'] > 2000]
filter3 = filter2[filter2['attacktype1_txt'] == 'Bombing/Explosion']  # Optional: focus on bombings

final_filter = filter3.groupby(['latitude','longitude']).size().reset_index(name='incident_count')

heatmap_data = final_filter[['latitude','longitude','incident_count']].values.tolist()

casualty_map = folium.Map(location=[20, 0], zoom_start=2)
HeatMap(heatmap_data, radius=8).add_to(casualty_map)

casualty_map.save('heatmap_casualties_deadly.html')
casualty_map
```

This heatmap visualizes locations where fatal terrorist attacks (with at least one person killed) occurred after the year 2000. It highlights the global reach of deadly incidents, revealing the regions with the highest numbers of fatalities due to terrorism.
